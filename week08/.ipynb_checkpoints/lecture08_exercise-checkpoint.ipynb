```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import json
import urllib.request
```

```python
# Task 1 — World choropleth: life expectancy diverging scale

gm = px.data.gapminder()
gm_2007 = gm[gm['year'] == 2007].copy()

global_avg = gm_2007['lifeExp'].mean()
gm_2007['lifeExp_deviation'] = gm_2007['lifeExp'] - global_avg

fig1 = px.choropleth(
    gm_2007,
    locations='iso_alpha',
    color='lifeExp_deviation',
    hover_name='country',
    hover_data={
        'lifeExp': ':.1f',
        'lifeExp_deviation': ':.1f',
        'iso_alpha': False
    },
    color_continuous_scale='RdBu',
    color_continuous_midpoint=0,
    labels={
        'lifeExp_deviation': 'Life expectancy vs global average',
        'lifeExp': 'Life expectancy'
    },
    title='Sub-Saharan Africa sits furthest below the global life expectancy average in 2007'
)

fig1.update_layout(
    template='plotly_white',
    font=dict(family='Arial', size=12),
    height=650,
    margin=dict(l=20, r=20, t=80, b=20)
)

fig1.show()
```

### Task 2 — Design decisions

**GeoJSON source:** GitHub datasets/geo-countries country polygons GeoJSON. This dataset provides GeoJSON country polygons suitable for global country-level mapping.

**Chart type chosen:** I used `px.choropleth` because this is a country-level map and I do not need street-level zoom or map tiles. A standard choropleth is clearer and simpler for comparing countries.

**Colour scale chosen:** I used a sequential colour scale because the variable is GDP per capita. Higher values represent more of the same quantity, so a light-to-dark sequential scale is appropriate.

```python
# Task 2 — GeoJSON choropleth using country polygons

geojson_url = "https://raw.githubusercontent.com/datasets/geo-countries/master/data/countries.geojson"

with urllib.request.urlopen(geojson_url) as response:
    my_geojson = json.load(response)

print(my_geojson['features'][0]['properties'])
```

```python
gm = px.data.gapminder()
gm_2007 = gm[gm['year'] == 2007].copy()

fig2 = px.choropleth(
    gm_2007,
    geojson=my_geojson,
    locations='iso_alpha',
    featureidkey='properties.ISO3166-1-Alpha-3',
    color='gdpPercap',
    hover_name='country',
    hover_data={
        'gdpPercap': ':.0f',
        'lifeExp': ':.1f',
        'pop': ':,',
        'iso_alpha': False
    },
    color_continuous_scale='Blues',
    labels={'gdpPercap': 'GDP per capita'},
    title='High GDP per capita is concentrated in North America, Western Europe, and Oceania'
)

fig2.update_geos(
    fitbounds='locations',
    visible=False
)

fig2.update_layout(
    template='plotly_white',
    font=dict(family='Arial', size=12),
    height=650,
    margin=dict(l=20, r=20, t=80, b=20)
)

fig2.show()
```
